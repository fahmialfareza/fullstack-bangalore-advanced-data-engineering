```python
import pandas as pd 
import sqlalchemy as sa
from datetime import datetime, timedelta, timezone

# Setting agar kolom bisa tampil semua
pd.set_option('display.max_columns', None)

# Proses ekstraksi data di gsheet
def extract_data_from_gsheet(sheet_id):
    url = f'https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=xlsx'
    data = pd.read_excel(url)
    return(data)

# Proses menambahkan metadata
def add_metadata(data):
    # Get date now
    gmt_plus_7 = timezone(timedelta(hours=14))
    current_time = datetime.now(gmt_plus_7)

    # Tambahkan kolom created date dan isikan dengan tanggal dan waktu hari ini
    data['created_date'] = current_time
    data['created_by'] = 'SYSTEM'
    return(data)

# Proses get connection to database
def connect_to_database():
    username = 'airflow'
    password = 'airflow'
    localhost = 'host.docker.internal'
    port = '5433'
    database = 'db_pizza_staging'
    
    engine = sa.create_engine(
        f"postgresql+psycopg://{username}:{password}@{localhost}:{port}/{database}"
    )
    connection = engine.connect()

    return(engine, connection)

# Proses load data / menyimpan data ke database
def load_data(data, table_name, connection):
    data.to_sql(
        table_name,
        connection,
        if_exists = 'replace',
        index = False
    )
```

```python
# Buat pemetaan data dan sheet id
mapping_url = {
    'trx_order' : '1mQpORyAxRPOC0lPE8eh35H5KxEZOsFBN',
    'trx_order_detail' : '1s6nM3BA__FuMnTmR-SIMHPTHj3HLbtb1',
    'mst_pizza' : '1ORuqaWkWMaJYAwBWJrfvC-65xPaQ-Y5F',
    'mst_pizza_type' : '1uDoShPx2Fa-MOeXMac0WmENfNgUZeR3K',
    'mst_customer' : '12jXvSGYxBIT9Q_iPq2z4KMpfRLr7m0aP'
}

# Buat koneksi ke database
engine, conn = connect_to_database()

# Proses ETL Data dengan looping
for key, value in mapping_url.items():
    print(f'Ekstrak data {key}')
    # Ekstrak data (E)
    data_extracted = extract_data_from_gsheet(sheet_id = value)
    
    # Add metadata
    data_extracted = add_metadata(data_extracted)
    display(data_extracted.head(2))
    print()
    
    # Load Data
    load_data(
        data = data_extracted, 
        table_name = key, 
        connection = conn
    )

    print(f'Proses ETL Data {key} berhasil!\n')
```
