```python
import sys
import pytz
import pandas as pd
import sqlalchemy as sa
from datetime import datetime, timezone

def extract_data(
    query : str, 
    connection : sa.engine.Connection
) -> pd.DataFrame:
    # Proses ekstrak data dari database
    data = pd.read_sql(f'{query}', connection)
    return(data)

def add_metadata(data : pd.DataFrame):
    # Atur/Sesuaikan Zona Waktu 
    now_utc = datetime.now(timezone.utc)
    date_now = now_utc.astimezone(pytz.timezone('Asia/Jakarta'))
    
    # Tambahkan dua buah kolom metadata (job_created_date & job_created_by)
    data['job_created_date'] = date_now
    data['job_created_by'] = 'SYSTEM'
    return(data)

def create_db_connection(**params : dict):
    hostname = params.get('hostname')
    portname = params.get('portname')
    username = params.get('username')
    password = params.get('password')
    database = params.get('database')

    url = (
        f'postgresql+psycopg2://'
        f'{username}:{password}@{hostname}:{portname}/{database}'
    )

    engine = sa.create_engine(url)
    conn = engine.connect()
    return (engine, conn)

def load_data(
    data : pd.DataFrame, 
    connection : sa.engine.Connection,
    tbl_destination_name : str
):
    data.to_sql(
        tbl_destination_name, 
        connection, 
        if_exists = 'replace', 
        index = False
    ) 
    
    connection.close()
```

```python
query_integration = """ 
    WITH tbl_order AS (
    	SELECT 
    		order_id,
    		customer_id,
    		order_maker_id,
    		CAST(TO_CHAR(TO_DATE(order_date, 'MM/DD/YYYY'), 'YYYY-MM-DD') AS DATE) as order_date,
    		order_time, 
    		completion_time,
    		TO_CHAR(completion_time::TIME - order_time::TIME, 'MI:SS') AS processing_time,
    		is_complain,
    		complain_detail
    	FROM trx_order
    	WHERE order_id IS NOT NULL
    ),
    tbl_order_detail AS (
    	SELECT 
    		order_details_id,
    		order_id,
    		pizza_id,
    		quantity
    	FROM trx_order_detail
    	WHERE order_details_id IS NOT NULL
    ),
    tbl_pizza AS (
    	SELECT DISTINCT
    		pizza_id, 
    		pizza_type_id, 
    		size, 
    		CAST(REPLACE(price, 'IDR', '') AS FLOAT) AS price,
    		CAST(REPLACE(production_cost, 'IDR', '') AS FLOAT) AS production_cost
    	FROM mst_pizza
    	WHERE pizza_id IS NOT NULL
    ),
    tbl_pizza_type AS (
    	SELECT 
    		pizza_type_id,
    		UPPER(name) AS pizza_name,
    		UPPER(category) AS pizza_category,
    		ingredients
    	FROM mst_pizza_type 
    	WHERE pizza_type_id IS NOT NULL
    ),
    tbl_customer as (
    	SELECT 
    		customer_id,
    		customer_name, 
    		gender as customer_gender,
    		birth_date as customer_birth_date, 
    		phone_number as customer_phone_no,
    		email as customer_email,
    		CAST(TO_CHAR(TO_DATE(open_date, 'YYYY/MM/DD'), 'YYYY-MM-DD') AS DATE) as customer_open_date
    	FROM mst_customer 
    	WHERE customer_id IS NOT NULL
    )
    	SELECT
    		o.order_id,
    		od.order_details_id,
    		o.customer_id,
    		o.order_maker_id,
    		od.pizza_id,
    		pt.pizza_type_id,
    		o.order_date,
    		o.order_time, 
    		o.completion_time,
    		o.processing_time,
    		o.is_complain,
    		o.complain_detail,
    		od.quantity,
    		p.size, 
    		p.price,
    		p.production_cost,
    		od.quantity * p.price AS total_price,
    		od.quantity * (p.price - p.production_cost) AS profit,
    		pt.pizza_name,
    		pt.pizza_category,
    		pt.ingredients,
    		c.customer_id,
    		c.customer_name, 
    		c.customer_gender,
    		c.customer_birth_date, 
    		c.customer_phone_no,
    		c.customer_email,
    		c.customer_open_date
    	FROM tbl_order o 
    	INNER JOIN tbl_order_detail od on o.order_id = od.order_id 
    	LEFT JOIN tbl_pizza p ON od.pizza_id = p.pizza_id 
    	LEFT JOIN tbl_pizza_type pt ON p.pizza_type_id = pt.pizza_type_id 
    	LEFT JOIN tbl_customer c on o.customer_id = c.customer_id 
"""
```

```python
engine_src, conn_src = create_db_connection(
    hostname = 'host.docker.internal',
    portname = '5433',
    username = 'airflow',
    password = 'airflow',
    database = 'db_stagging'
)

data_extracted = extract_data(query = query_integration, connection = conn_src)
data_extracted = add_metadata(data_extracted)
display(data_extracted)
```
