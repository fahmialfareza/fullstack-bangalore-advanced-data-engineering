```python
import pandas as pd 
import sqlalchemy as sa

# Setting agar column bisa tampil semua
pd.set_option('display.max_columns', None)
```

```python
# Proses ekstraksi data
url = 'https://raw.githubusercontent.com/Pradnya1208/Telecom-Customer-Churn-prediction/refs/heads/main/data.csv'
data_customer_churn = pd.read_csv(url)

display(data_customer_churn)
```

```python
# Cek informasi umum pada data
data_customer_churn.info()
```

```python
# Akses nama column
data_customer_churn.columns
```

```python
# Konversi nama kolom dari PascalCase menjadi lowercase
data_customer_churn.columns = data_customer_churn.columns.str.lower()
display(data_customer_churn.columns)
```

```python
display(data_customer_churn)
```

```python
from datetime import datetime, timedelta, timezone

# Get date now
gmt_plus_7 = timezone(timedelta(hours=7))
current_time = datetime.now(gmt_plus_7)

# Tambahkan kolom created date dan isikan dengan tanggal dan waktu hari ini
data_customer_churn['created_date'] = current_time
data_customer_churn.head()
```

```python
# Proses pembuatan koneksi ke database
import sqlalchemy as sa

username = 'airflow'
password = 'airflow'
localhost = 'postgres-db'
port = '5432'
database = 'db_staging'

engine = sa.create_engine(
    f"postgresql+psycopg://{username}:{password}@{localhost}:{port}/{database}"
)
connection = engine.connect()
```

```python
data_customer_churn.to_sql(
    'customer_churn',
    connection,
    if_exists = 'replace',
    index = False
)
```
