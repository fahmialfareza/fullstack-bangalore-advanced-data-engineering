```python
import sys
import pytz
import pandas as pd
import sqlalchemy as sa
from datetime import datetime, timezone

def extract_data(sheet_id : str):
    # Proses ekstrak data dari google sheet
    url = f'https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=xlsx'
    data = pd.read_excel(url)
    return(data)

def add_metadata(data : pd.DataFrame):
    # Atur/Sesuaikan Zona Waktu 
    now_utc = datetime.now(timezone.utc)
    date_now = now_utc.astimezone(pytz.timezone('Asia/Jakarta'))
    
    # Tambahkan dua buah kolom metadata (job_created_date & job_created_by)
    data['job_created_date'] = date_now
    data['job_created_by'] = 'SYSTEM'
    return(data)

def create_db_connection(**params : dict):
    hostname = params.get('hostname')
    portname = params.get('portname')
    username = params.get('username')
    password = params.get('password')
    database = params.get('database')

    url = (
        f'postgresql+psycopg2://'
        f'{username}:{password}@{hostname}:{portname}/{database}'
    )

    engine = sa.create_engine(url)
    conn = engine.connect()
    return (engine, conn)

def load_data(
    data : pd.DataFrame, 
    connection : sa.engine.Connection,
    tbl_destination_name : str
):
    data.to_sql(
        tbl_destination_name, 
        connection, 
        if_exists = 'replace', 
        index = False
    ) 
    
    connection.close()

if __name__ == '__main__': 
    sheet_id = '1UCbczE9AiD77zrj9Qa4eR2N6XQ0GOm4o'
    data_id = extract_data(sheet_id)
    data_name = sys.argv[1]

    condition = data_id['data_name'] == data_name
    selected_sheet_id = data_id.loc[condition]['sheet_id'].values[0]

    # Proses ekstrak data
    data_extracted = extract_data(sheet_id = selected_sheet_id)
    data_extracted = add_metadata(data_extracted)

    # Proses pembuatan koneksi db
    engine_dest, conn_dest = create_db_connection(
        hostname = 'host.docker.internal',
        portname = '5433',
        username = 'airflow',
        password = 'airflow',
        database = 'db_stagging'
    )

    load_data(
        data = data_extracted, 
        connection = conn_dest,
        tbl_destination_name = data_name
    )

    print(f'ETL data {data_name} sukses')
```
